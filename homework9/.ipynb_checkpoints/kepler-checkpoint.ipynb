```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
```

```python
df = pd.read_csv("../../course_assignments/homework9/solar_system.csv")
print(df.head())
print(df.shape)
#rectangle with 20 rows and 11 cols
print(df.tail())
```

```python
print(df.columns)
print(df.index)
```

```python
orbital_period = df.iloc[10].astype(float)
perihelion = df.iloc[8].astype(float)
aphelion = df.iloc[9].astype(float)

print(orbital_period)
#amount of time it takes for the celestial object to orbit its sun
print(perihelion)
#distance closest to its sun in orbit
print(aphelion)
#distace farthest to its sun in orbit
```

```python
semi_major_axis = (perihelion + aphelion) / 2
new_row = pd.Series(semi_major_axis)
#df = pd.concat(df)

print(df.shape)
```

```python
plt.scatter(semi_major_axis, orbital_period, label="solar system bodies")

plt.xlabel("Semi-major axis (10^6 km)")
plt.ylabel("Orbital period (days)")
plt.title("Semi-Major Axis vs. Orbital Period")
plt.legend()

plt.show()
```

```python
def keplers_third_law(a, m):
  return m * a**1.5

popt, pcov = curve_fit(keplers_third_law, semi_major_axis, orbital_period)

m_fit = popt[0]

print(f"Fitted model: T = {m_fit:.2f} * a^1.5")
```

```python
y_pred = keplers_third_law(semi_major_axis, m_fit)

residuals = orbital_period - y_pred
chi2 = np.sum(residuals**2)

dof = len(semi_major_axis) - len(popt)
chi2_reduced = chi2 / dof

print(chi2_reduced)
```

```python
a_smooth = np.linspace(semi_major_axis.min(), semi_major_axis.max(), 200)
period_fit = keplers_third_law(a_smooth, m_fit)

plt.scatter(semi_major_axis, orbital_period, color="blue", label="Data")
plt.plot(a_smooth, period_fit, color="red",)

plt.xlabel("Semi-major axis (10^6 km)")
plt.ylabel("Orbital period (days)")
plt.title(f"Kepler's Third Law Fit | Reduced Chi-Squared = {chi2_reduced:.2f}")
plt.legend()

plt.show()
```
