```python
import numpy as np
import matplotlib.pyplot as plt

def make_sine_wave(x, A, w):
    return A * np.sin(w * x)

x = np.linspace(0, 2 * np.pi, 1000)
amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))

for A, w in zip(amplitude, frequency):
    y1 = make_sine_wave(x, A, 1)
    y2 = make_sine_wave(x, 1, w)
    y3 = make_sine_wave(x, A, w)

    axes[0].plot(x, y1, linestyle='-')
    axes[1].plot(x, y2, linestyle='-.')
    axes[2].plot(x, y3, linestyle='--')

axes[0].set_title("Varying Amplitude")
axes[0].set_xlabel("time (s)")
axes[0].set_ylabel("amps (m)")

axes[1].set_title("Varying Frequency")
axes[1].set_xlabel("time (s)")
axes[1].set_ylabel("amps (m)")

axes[2].set_title("Varying Amplitude and Frequency")
axes[2].set_xlabel("time (s)")
axes[2].set_ylabel("amps (m)")


plt.show()
```
