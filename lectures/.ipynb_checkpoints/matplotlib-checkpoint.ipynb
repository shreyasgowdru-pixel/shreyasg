```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# sin function

x = np.linspace(0, 2*np.pi, 100)
y = np.sin(x)
```

```python
plt.figure(figsize = (6,4))
plt. plot(x, y)
plt.show()
```

```python
y1 = np.sin(x)
y2 = np.cos(x)
```

```python
plt.figure()

plt.scatter(x, y1, label = "sin(x)", color = "red")
plt.scatter(x, y2, label = "cos(x)", color = "pink")

plt.title("sine and cosine wave")
plt.xlabel("x values (radians)")
plt.ylabel("f(x)")

plt.legend()
plt.show()
```

```python
# customization

plt.figure()
plt.plot(x, y1, color = "xkcd:icky green", linestyle = "-.")

plt.show()
```

```python
# subplots

fig, axes = plt.subplots(1, 2, figsize = (10, 4))

axes[0].plot(x, y1, color = "plum", ls = ":")

axes[0].set_title("Sine Wave", color = "darkorchid")

axes[1].plot(x, y2, color = "mediumslateblue", ls = "--")

axes[1].set_title("Cosine Wave", color = "darkslateblue")

plt.show()
```
