```python
arr1 = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13]
])

def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False
    return True
    
def return_row(arr):
    rows = []
    for row in arr:
        row_prime = False
        for el in row:
            if is_prime(el):
                row_prime = True
                break
        if row_prime:
            rows.append(row)
    return np.vstack(rows)

print(return_row(arr1))
```

```python
board = np.zeros((8,8), dtype=int)

def checkerboard(board):
    board[0::2, 1::2] = 1
    board[1::2, 0::2] = 1
    return board

print(checkerboard(board))
```

```python
def expansion(string, number):
    new_string = ""
    space = " " * number

    for i in range(len(string)):
        new_string += string[i]
        if i < len(string)-1:
            new_string += space
    return new_string

print(expansion("interstellar", 2))
```

```python
np.random.seed(1)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
```

```python
# note the smaller the number the brighter the star

def secondDimmest(stars):
    dim_stars = np.array([])
    for row in stars:
        brightest_in_row = row[0]
        second_brightest = row[0]
        for element in row:
            if element < brightest_in_row:
                second_brightest = brightest_in_row
                brightest_in_row = element
            elif element < second_brightest:
                second_brightest = element
        dim_stars = np.append(dim_stars, second_brightest)
    return dim_stars
print(secondDimmest(stars))
```
